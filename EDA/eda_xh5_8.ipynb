```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Extract sentiment (x), fans (y), and IP group for coloring
sentiment = df['component_scores_dim_3'].apply(
    lambda x: x.get('score3c_sentiment', np.nan) if isinstance(x, dict) else np.nan
)
fans = df['fans']
ip_group = df['ip_location'].apply(
    lambda loc: 'US' if isinstance(loc, str) and ('美国' in loc or 'United States' in loc) else 'Other'
)

# 2. Filter out missing values
mask = sentiment.notna() & fans.notna()
x = sentiment[mask].values.reshape(-1, 1)
y = fans[mask].values
groups = ip_group[mask]

# 3. Fit a linear regression model: fans = a * sentiment + b
model = LinearRegression().fit(x, y)
a, b = model.coef_[0], model.intercept_
r2 = model.score(x, y)

# 4. Prepare regression line
line_x = np.linspace(x.min(), x.max(), 200)
line_y = a * line_x + b

# 5. Plot scatter colored by IP group
plt.figure(figsize=(8, 6))
for grp, color in zip(['US', 'Other'], ['steelblue', 'lightgreen']):
    idx = (groups == grp)
    plt.scatter(x[idx], y[idx], s=50, alpha=0.6,
                label=f'{grp}', color=color)

# 6. Plot regression line
plt.plot(line_x, line_y, 'r--', linewidth=2,
         label=f'Fit: y = {a:.2f}·x + {b:.1f}')

# 7. Decorations
plt.title('Number of Fans vs Sentiment Score')
plt.xlabel('Sentiment Score (score3c_sentiment)')
plt.ylabel('Number of Fans')
# plt.text(0.05, 0.95, f'$R^2$ = {r2:.2f}',
#          transform=plt.gca().transAxes,
#          va='top',
#          bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))
plt.legend(title='IP Group')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Build df_plot from existing df
df_plot = pd.DataFrame({
    'ip_location': df['ip_location'],
    'follows': df['follows'],
    'fans': df['fans'],
    'interaction': df['interaction'],
    'score3c_sentiment': df['component_scores_dim_3']
        .apply(lambda x: x.get('score3c_sentiment', np.nan) if isinstance(x, dict) else np.nan)
}).dropna()

# Helper function: group into equal-width bins and label by integer midpoint
def plot_integer_binned_mean(series, sentiment, bins, title, xlabel):
    binned = pd.cut(series, bins=bins)
    means = df_plot.groupby(binned)[sentiment].mean()
    mids = [(interval.left + interval.right) / 2 for interval in means.index]
    labels = [int(mid) for mid in mids]
    # Sort by label
    labels, vals = zip(*sorted(zip(labels, means.values)))
    width = (labels[1] - labels[0]) * 0.8
    plt.figure(figsize=(10, 4))
    plt.bar(labels, vals, width=width, edgecolor='black', alpha=0.7)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Mean score3c_sentiment')
    plt.xticks(labels, rotation=45)
    plt.tight_layout()
    plt.show()

# Sentiment vs follows
plot_integer_binned_mean(
    df_plot['follows'], 'score3c_sentiment', 10,
    'Mean sentiment by Follows (10 bins)', 'Follows (midpoint)'
)

# Sentiment vs fans
plot_integer_binned_mean(
    df_plot['fans'], 'score3c_sentiment', 10,
    'Mean sentiment by Fans (10 bins)', 'Fans (midpoint)'
)

# Sentiment vs interaction
plot_integer_binned_mean(
    df_plot['interaction'], 'score3c_sentiment', 10,
    'Mean sentiment by Interaction (10 bins)', 'Interaction (midpoint)'
)

# IP Location bar chart (unchanged)
df_plot['loc_group'] = df_plot['ip_location'].apply(
    lambda x: 'US' if '美国' in str(x) or 'United States' in str(x) else 'Other'
)
mean_loc = df_plot.groupby('loc_group')['score3c_sentiment'].mean()
plt.figure(figsize=(6, 4))
plt.bar(mean_loc.index, mean_loc.values, edgecolor='black', alpha=0.7)
plt.title('Mean sentiment by IP Location')
plt.xlabel('IP Location')
plt.ylabel('Mean score3c_sentiment')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1) Robustly parse the age‐distribution columns, treating "Null" (any case) as NaN:
age_cols = ["粉丝年龄<18", "粉丝年龄18-24", "粉丝年龄25-34", "粉丝年龄35-44", "粉丝年龄>44"]

def parse_pct(x):
    if pd.isnull(x):
        return np.nan
    s = str(x).strip()
    if s.lower() == 'null' or s == '':
        return np.nan
    # replace Chinese comma
    s = s.replace('，', '.')
    try:
        return float(s)
    except ValueError:
        return np.nan

for col in age_cols:
    df[col] = df[col].apply(parse_pct)

# 2) Bar chart: average age distribution across all users
avg_dist = df[age_cols].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_dist.index, avg_dist.values, edgecolor='black', alpha=0.7)
plt.title("Average Followers Age Distribution")
plt.xlabel("Age Group")
plt.ylabel("Average Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3) Radar chart: age distribution for top 5 users by fans
top5 = df.nlargest(5, "fans")
labels = ["<18", "18-24", "25-34", "35-44", ">44"]
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # close the loop

plt.figure(figsize=(6,6))
ax = plt.subplot(polar=True)
for _, row in top5.iterrows():
    values = [row[col] for col in age_cols]
    values += values[:1]
    ax.plot(angles, values, marker='o', label=row['user_id'])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels)
ax.set_rlabel_position(30)
plt.title("Age Distribution for Top 5 Users by Fans", y=1.1)
plt.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ensure df is loaded with the necessary columns.

# 1. Parse percentages into numeric form
def parse_pct(x):
    try:
        s = str(x).replace('%', '').replace('，', '.').strip()
        return float(s)
    except:
        return np.nan

# Age group columns
age_cols = ["粉丝年龄<18", "粉丝年龄18-24", "粉丝年龄25-34", "粉丝年龄35-44", "粉丝年龄>44"]
for col in age_cols:
    df[col] = df[col].apply(parse_pct)

# Female ratio
df['female_pct'] = df['粉丝女性比例'].apply(parse_pct)

# Overseas pct
def get_overseas_pct(region_str):
    if not isinstance(region_str, str):
        return np.nan
    for line in region_str.split('\n'):
        parts = line.split('\t')
        if len(parts)>=3 and parts[1].strip()=='海外':
            return parse_pct(parts[2])
    return np.nan

df['overseas_pct'] = df['地域分布'].apply(get_overseas_pct)

# 2. Select top 3 users by fans
top3 = df.nlargest(10, 'fans')

# 3. Radar chart axes
labels = ['Age<18','18-24','25-34','35-44','>44','Female%','Overseas%']
num_vars = len(labels)
angles = np.linspace(0, 2*np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

# 4. Plot radar for each user
plt.figure(figsize=(6,6))
ax = plt.subplot(polar=True)

for _, row in top3.iterrows():
    values = [row[col] for col in age_cols] + [row['female_pct'], row['overseas_pct']]
    values = [0 if np.isnan(v) else v for v in values]
    values += values[:1]
    ax.plot(angles, values, marker='o', label=row['user_id'])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels)
ax.set_rlabel_position(30)
plt.title('Fan Demographics: Top 10 Users', y=1.1)
plt.legend(loc='upper right', bbox_to_anchor=(1.3,1.1))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Ensure the working DataFrame has the necessary columns
# Extract overseas_pct if not already present
def get_overseas_pct(region_str):
    if not isinstance(region_str, str):
        return np.nan
    for line in region_str.split('\n'):
        parts = line.split('\t')
        if len(parts) >= 3 and parts[1].strip() == '海外':
            try:
                return float(parts[2].strip().replace('%',''))
            except:
                return np.nan
    return np.nan

if 'overseas_pct' not in df:
    df['overseas_pct'] = df['地域分布'].apply(get_overseas_pct)

# Extract sentiment score if not present
if 'score3c_sentiment' not in df:
    df['score3c_sentiment'] = df['component_scores_dim_3'].apply(
        lambda x: x.get('score3c_sentiment', np.nan) if isinstance(x, dict) else np.nan
    )

# Prepare subset
plot_df = df.dropna(subset=['overseas_pct', 'score3c_sentiment', 'fans', 'ip_location'])

# Create ip_group
plot_df['ip_group'] = plot_df['ip_location'].apply(
    lambda x: 'US' if '美国' in str(x) or 'United States' in str(x) else 'Other'
)

# 1. Scatter: sentiment vs overseas %, bubble size ~ fans, color = ip_group
sizes = (plot_df['fans'] / plot_df['fans'].max()) * 200
colors = plot_df['ip_group'].map({'US':'skyblue','Other':'lightgreen'})
plt.figure(figsize=(8,6))
for grp in ['US','Other']:
    subset = plot_df[plot_df['ip_group']==grp]
    plt.scatter(subset['overseas_pct'], subset['score3c_sentiment'],
                s=(subset['fans']/subset['fans'].max())*200,
                alpha=0.6, label=grp)
plt.xlabel('Overseas Fan %')
plt.ylabel('Sentiment Score')
plt.title('Sentiment vs Overseas Fan % (bubble ∝ fans)')
plt.legend(title='IP Group')
plt.grid(True)
plt.show()

# 2. Bar: mean sentiment by 10-point overseas_pct bins
bins = np.arange(0, 101, 10)
plot_df['over_bin'] = pd.cut(plot_df['overseas_pct'], bins)
mean_sent = plot_df.groupby('over_bin')['score3c_sentiment'].mean()
labels = [f'{int(interval.left)}–{int(interval.right)}%' for interval in mean_sent.index]
plt.figure(figsize=(10,4))
plt.bar(labels, mean_sent.values, edgecolor='black', alpha=0.7)
plt.title('Average Sentiment by Overseas % Range')
plt.xlabel('Overseas Fan % Range')
plt.ylabel('Mean Sentiment Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
