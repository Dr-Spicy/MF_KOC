# 1.互动效能

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Distribution with quartiles and insight
plt.figure(figsize=(10, 6))
sns.histplot(df_flat['component_scores.score1a_efficiency'], bins=30, kde=True, color='#4F8FC5')
q1, q2, q3 = np.percentile(df_flat['component_scores.score1a_efficiency'].dropna(), [25, 50, 75])
plt.axvline(q1, color='orange', linestyle='--', label='25th percentile')
plt.axvline(q2, color='green', linestyle='--', label='Median')
plt.axvline(q3, color='red', linestyle='--', label='75th percentile')
plt.title('Distribution of Interaction Efficiency')
plt.xlabel('Interaction Efficiency Score')
plt.ylabel('Number of Creators')
plt.legend()
plt.suptitle('Most creators cluster at the lower end, but top quartile may have outsized impact. \nConsider targeting top 25% for brand collaborations.', fontsize=10, color='gray', y=0.92)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='component_scores.score1a_efficiency', y='fans', data=df_flat, alpha=0.5, color='#C2544F')
sns.regplot(x='component_scores.score1a_efficiency', y='fans', data=df_flat, scatter=False, color='black', line_kws={'linestyle':'dashed'})
plt.title('Interaction Efficiency vs. Number of Fans')
plt.xlabel('Interaction Efficiency Score')
plt.ylabel('Number of Fans')
plt.suptitle('Top-right corner = high engagement AND big fanbase. Ideal KOC partners!', fontsize=10, color='gray', y=0.92)
# # Highlight a few top performers (optional)
# top_creators = df_flat.nlargest(3, ['互动效能', 'fans'])
# for i, row in top_creators.iterrows():
#     plt.annotate(row['nickname'], (row['互动效能'], row['fans']), xytext=(5, 10), textcoords='offset points', fontsize=8, color='blue')
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

# 2.账号活跃度

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Calculate how many creators share the same (账号活跃度, 互动效能) value
df_flat['point_group'] = list(zip(df_flat['component_scores.score1c_activity'], df_flat['component_scores.score1a_efficiency']))
point_counts = df_flat['point_group'].value_counts().to_dict()
df_flat['point_count'] = df_flat['point_group'].map(point_counts)

# Make a scatterplot with point size proportional to how many creators share the position
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='component_scores.score1c_activity',
    y='component_scores.score1a_efficiency',
    data=df_flat,
    size='point_count',  # key: size by group count
    sizes=(50, 300),     # adjust min/max size as needed
    alpha=0.6,
    color='#10b981',
    legend=False
)
sns.regplot(
    x='component_scores.score1c_activity',
    y='component_scores.score1a_efficiency',
    data=df_flat,
    scatter=False,
    color='black',
    line_kws={'linestyle': 'dashed'}
)
plt.title('Account Activity vs. Interaction Efficiency', fontsize=15)
plt.xlabel('Account Activity Score', fontsize=12)
plt.ylabel('Interaction Efficiency Score', fontsize=12)
plt.tight_layout()
plt.show()


```

```python
top_kocs = df_flat.sort_values(['component_scores.score1c_activity', 'component_scores.score1a_efficiency', 'fans'], ascending=False).head(10)
display_cols = ['nickname', 'fans', 'component_scores.score1c_activity', 'component_scores.score1a_efficiency', 'note_count', 'location', 'total_score']
top_kocs[display_cols]
```
