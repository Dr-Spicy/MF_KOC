# 1. 兴趣标签匹配度

["美食", "生活记录", "探店", "素材"]

```python
def compute_interest_score(tag_str, target_keywords=None):
    """
    Compute the matching score for the '兴趣标签' column based on bidirectional substring matching,
    and return it rounded to one decimal place.

    Parameters:
        tag_str (str): The cell string containing multiple interest tags,
                       expected in the format "序号\t标签\t百分比" per line.
        target_keywords (list): List of keywords to match. Defaults to ["美食", "生活记录", "探店", "素材"].

    Returns:
        float: The computed score (max 40), rounded to 0.1.
    """
    if target_keywords is None:
        target_keywords = ["美食", "生活记录", "探店", "素材"]

    if not isinstance(tag_str, str):
        return np.nan

    total_pct = 0.0
    for line in tag_str.split('\n'):
        parts = line.split('\t')
        if len(parts) >= 3:
            tag = parts[1].strip()
            try:
                pct = float(parts[2].strip().replace('%', ''))
            except ValueError:
                pct = 0.0
            for kw in target_keywords:
                if kw in tag or tag in kw:
                    total_pct += pct
                    break

    raw = 40.0 if total_pct >= 25 else (total_pct / 25 * 50)
    return round(raw, 1)

def add_interest_score_column(df):
    """
    Apply compute_interest_score to the dataframe and add a new column
    'Interest tag matching score' with one-decimal precision.
    """
    df = df.copy()
    df['score4a_tag'] = df['兴趣标签'].apply(compute_interest_score)
    return df
```

# 2. 地域聚焦度

## 粉丝IP地址海外比例逆向阶梯模型

```python
import pandas as pd
import numpy as np

def compute_overseas_score(region_str):
    """
    Compute the score based on the overseas (海外) proportion using a reverse step model,
    and return it rounded to one decimal place.

    - If the '海外' percentage is ≥ 50%: score is 10.0
    - For each 10% decrease below 50, subtract 2.0 points
    - If '海外' is not mentioned or is below 10%, score is 0.0

    Parameters:
        region_str (str): The string from the '地域分布' column.

    Returns:
        float: The computed overseas score, rounded to 0.1.
    """
    if not isinstance(region_str, str):
        return np.nan

    overseas_pct = None
    for line in region_str.split('\n'):
        parts = line.split('\t')
        if len(parts) >= 3 and parts[1].strip() == "海外":
            try:
                overseas_pct = float(parts[2].strip().replace('%', ''))
            except ValueError:
                overseas_pct = 0.0
            break

    # Determine raw score
    if overseas_pct is None or overseas_pct < 10:
        raw = 0.0
    elif overseas_pct >= 50:
        raw = 10.0
    elif overseas_pct >= 40:
        raw = 8.0
    elif overseas_pct >= 30:
        raw = 6.0
    elif overseas_pct >= 20:
        raw = 4.0
    else:  # >=10 and <20
        raw = 2.0

    return round(raw, 1)

def add_overseas_score_column(df):
    """
    Apply compute_overseas_score to the dataframe and add a new column 'Fan IP address overseas score'
    with one-decimal precision.
    """
    df = df.copy()
    df['Fan IP address overseas score'] = df['地域分布'].apply(compute_overseas_score)
    return df
```

```python
import re
import numpy as np

def compute_geo_keyword_score(fan_overseas_score, posts, B_geo_cover=1.0, geo_keywords=None):
    """
    Compute the final score combining the fan overseas score and the geographic keyword coverage score,
    rounded to one decimal place.
    """
    if geo_keywords is None:
        geo_keywords = ["DFW", "Plano", "Dallas", "Fort Worth", "Arlington",
                        "达拉斯", "75023", "75024", "75025", "75026", "75074",
                        "75075", "75086", "75093", "75094"]

    if not isinstance(posts, list):
        posts = [posts]

    total_posts = len(posts)
    if total_posts == 0:
        geo_coverage_ratio = 0.0
    else:
        patterns = [re.compile(re.escape(k), re.IGNORECASE) for k in geo_keywords]
        matching = sum(any(p.search(post) for p in patterns) for post in posts)
        geo_coverage_ratio = matching / total_posts

    # Compute the raw geographic component, then clamp to [0, 10]
    raw_geo = (10 * geo_coverage_ratio) / B_geo_cover
    geo_score = min(max(raw_geo, 0.0), 10.0)

    # Final score is fan overseas + clamped geo component, rounded to 0.1
    return round(fan_overseas_score + geo_score, 1)

def add_geo_keyword_score_to_df(df, df_contents, B_geo_cover=1.0, geo_keywords=None, post_col='desc'):
    """
    Apply compute_geo_keyword_score and add a new column 'Geographic keyword coverage score'
    with one-decimal precision.
    """
    posts_by_user = df_contents.groupby('user_id')[post_col].apply(list).to_dict()
    df = df.copy()

    def compute_user_geo_score(row):
        posts = posts_by_user.get(row['user_id'], [])
        return compute_geo_keyword_score(
            row.get('Fan IP address overseas score', np.nan),
            posts,
            B_geo_cover,
            geo_keywords
        )

    df['Geographic keyword coverage score'] = df.apply(compute_user_geo_score, axis=1)
    return df
```

# 3. 粉丝年龄段匹配度

```python
def compute_age_matching_score(row, target_vector=np.array([10, 30, 30, 15, 15])):
    """
    Compute the age matching score for one user based on their follower age distribution,
    rounded to one decimal place.
    """
    age_cols = ["粉丝年龄<18", "粉丝年龄18-24", "粉丝年龄25-34", "粉丝年龄35-44", "粉丝年龄>44"]

    try:
        user_vector = np.array([float(str(row[col]).replace('，', '.')) for col in age_cols])
    except:
        return np.nan

    norm_user = np.linalg.norm(user_vector)
    norm_target = np.linalg.norm(target_vector)

    if norm_user == 0 or norm_target == 0:
        cosine_sim = 0.0
    else:
        cosine_sim = np.dot(user_vector, target_vector) / (norm_user * norm_target)

    score = 20 * ((cosine_sim + 1) / 2)
    return round(score, 1)

def add_age_matching_score(df):
    """
    Apply the age matching score computation to the DataFrame and add a new column
    'Fan age matching score' rounded to one decimal place.
    """
    df = df.copy()
    df['score4c_age'] = df.apply(compute_age_matching_score, axis=1)
    return df
```

# 4. 粉丝性别匹配度

```python
def compute_female_ratio_score(female_ratio, k=2, sigma=15):
    """
    Compute the dynamic compensation score based on the deviation of the fan female ratio from 80%,
    rounded to one decimal place.
    """
    # Check for missing values
    if female_ratio is None or (isinstance(female_ratio, float) and np.isnan(female_ratio)):
        return np.nan

    # If the value is a string, remove '%' and convert Chinese commas
    if isinstance(female_ratio, str):
        female_ratio = female_ratio.replace('%', '').replace('，', '.').strip()
        try:
            female_ratio = float(female_ratio)
        except ValueError:
            return np.nan

    # Compute absolute deviation from 80%
    deviation = abs(female_ratio - 80)

    # Compute the Laplace-like penalty score
    score = 10 * np.exp(-k * deviation / sigma)
    return round(score, 1)

def add_female_ratio_score(df, k=2, sigma=15):
    """
    Apply the dynamic compensation model for female ratio and add 'Fans gender matching score',
    rounded to one decimal place.
    """
    df = df.copy()
    df['score4d_gender'] = df['粉丝女性比例'].apply(
        lambda x: compute_female_ratio_score(x, k=k, sigma=sigma)
    )
    return df
```

```python
# Compute the new "dim4_score" column as the sum of the four component scores,
# then round to one decimal place in-place.
df['score4_target_match'] = (
    df['score4a_tag'] +
    df['score4b_region'] +
    df['score4c_age'] +
    df['score4d_gender']
).round(1)
```
